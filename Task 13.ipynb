# NumPy Data Filtering

## Aim
To filter numerical values using NumPy conditions.

## Description
This task demonstrates how to filter numerical data using different conditions and Boolean indexing.

## Objectives
- Create a NumPy array.
- Filter values greater than a threshold.
- Filter values less than a threshold.
- Select values based on conditions.

```python
import numpy as np

# Create a NumPy array
marks = np.array([20, 40, 50, 60, 80, 100, 35, 75, 45, 90])

# Display original data
print("Original Marks:", marks)

# Filter values greater than 50
print("Marks Above 50:", marks[marks > 50])

# Filter values less than 50
print("Marks Below 50:", marks[marks < 50])

# Filter values equal to 50
print("Marks Equal to 50:", marks[marks == 50])

# Filter values greater than or equal to 75
print("Marks 75 and Above:", marks[marks >= 75])

# Filter even numbers
print("Even Marks:", marks[marks % 2 == 0])
```

## Result

Successfully implemented NumPy data filtering using
Boolean indexing and different comparison conditions.
The program filters numerical values based on the
specified thresholds.
